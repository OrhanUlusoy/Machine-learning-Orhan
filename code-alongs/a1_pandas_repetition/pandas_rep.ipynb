```python
import sys

print("Python executable:", sys.executable)
print("Python prefix:", sys.prefix)

def _try_import(name: str):
    try:
        mod = __import__(name)
        version = getattr(mod, "__version__", "(no __version__)")
        print(f"OK: import {name} ({version})")
    except Exception as e:
        print(f"FAIL: import {name} -> {type(e).__name__}: {e}")

for pkg in ["pandas", "numpy", "matplotlib", "seaborn", "sklearn"]:
    _try_import(pkg)
```
